# Blurring

```python
img4 = cv2.imread("C://Users//ABDUL//Desktop//G&A_detect_Proj//PassPort.jpg")
img3=cv2.cvtColor(img4,cv2.COLOR_BGR2HSV) 
figure_size=7 
result=cv2.blur(img3, (figure_size, figure_size)) 
plt.figure(figsize=(11,6)) 
plt.subplot(121), plt.imshow(cv2.cvtColor(img3,cv2.COLOR_BGR2HSV)), plt.title("Original") 
plt.xticks([]), plt.yticks([]) 
plt.subplot(131), plt.imshow(cv2.cvtColor(result,cv2.COLOR_BGR2HSV)), plt.title("Blurred") 
plt.xticks([]), plt.yticks([]) 
plt.show
```

# Blurring


```python
img1 = cv2.imread("C://Users//ABDUL//Desktop//G&A_detect_Proj//fafda.png")
img=cv2.cvtColor(img1,cv2.COLOR_BGR2HSV) 
figure_size=7 
result=cv2.blur(img, (figure_size, figure_size)) 
plt.figure(figsize=(11,6)) 
plt.subplot(121), plt.imshow(cv2.cvtColor(img,cv2.COLOR_HSV2RGB)), plt.title("Original") 
plt.xticks([]), plt.yticks([]) 
plt.subplot(122), plt.imshow(cv2.cvtColor(result,cv2.COLOR_HSV2RGB)), plt.title("Blurred") 
plt.xticks([]), plt.yticks([]) 
plt.show()
```

# Zoom

```python
img1 = cv2.imread("C://Users//ABDUL//Desktop//G&A_detect_Proj//dp.jpg")
```

```python
(height,width)=img1.shape[0:2] 
zoom=cv2.resize(img1, None, fx=1.5,fy=1.5, interpolation=cv2.INTER_CUBIC) 
plt.imshow(zoom)
```

```python
zoom=cv2.resize(img1, (int(width*2), int (height*2)), interpolation=cv2.INTER_CUBIC) 
plt.imshow(zoom)
```
